````python
import torch
import torch.nn as nn
import torch.optim as optim
import copy

"""
MLP avec 2 couches cachées
"""
model = nn.Sequential(
    nn.Linear(in_features=61 , out_features=16), # 61 catactéristiques par molécules
    nn.ReLU(),

    nn.Linear(in_features=16 , out_features=8),
    nn.ReLU(),

    nn.Linear(in_features=8, out_features=1),
)
print(model)

"""
MLP avec 2 couches cachées et comparaisons avec SGD, SGD + Momentum, Adam
"""
lr_opt = 0.001

epochs = 10000


optimizers = { "SGD" : lambda m: optim.SGD(m.parameters(), lr_opt),
               "SGD + Momentum" : lambda m: optim.SGD(m.parameters(), lr_opt, momentum=0.9),
               "Adam" : lambda m: optim.Adam(model.parameters(), lr_opt)}
````

````python
# Vocabulaire : tous les caractères uniques présents dans les SMILES du jeu de données
vocab = sorted(set("".join(df["smiles"])))
char_to_idx = {c: i for i, c in enumerate(vocab)}


def smiles_to_features(smiles: str) -> np.ndarray:
    """Convertit une chaîne SMILES en vecteur de caractéristiques de longueur fixe.

    Le vecteur contient :
      - la fréquence de chaque caractère du vocabulaire (taille = len(vocab)),
      - la longueur de la chaîne SMILES.
    """
    counts = np.zeros(len(vocab), dtype=np.float32)
    for c in smiles:
        if c in char_to_idx:
            counts[char_to_idx[c]] += 1
    length = np.array([len(smiles)], dtype=np.float32)
    return np.concatenate([counts, length])


# Construire les matrices de caractéristiques
X_train = np.stack([smiles_to_features(s) for s in train_df["smiles"]])
X_val = np.stack([smiles_to_features(s) for s in val_df["smiles"]])
y_train = train_df["target"].values.astype(np.float32)
y_val = val_df["target"].values.astype(np.float32)

print(f"X_train : {X_train.shape}  (nb. molécules × nb. caractéristiques)")
print(f"X_val   : {X_val.shape}")
print(f"Caractéristiques par molécule : {X_train.shape[1]} ({len(vocab)} fréquences de caractères + 1 longueur)")
print(f"y_train : {y_train.shape}")

#Passage de numpy nd array en tenseur pytorch
X_tensor_train = torch.tensor(X_train, dtype=torch.float32)
X_tensor_val = torch.tensor(X_val, dtype=torch.float32)
y_tensor_train = torch.tensor(y_train).view(-1,1)
y_tensor_val = torch.tensor(y_val).view(-1,1)
````

# 1.1 MLP (V1)

````python
#Nombre d'époques
epochs = 10000
# Fonction de perte
criterion = nn.MSELoss()
lr = 0.001

model_v1 = copy.deepcopy(model)

for epoch in range(epochs):
  model_v1.train()
  predictions = model_v1(X_tensor_train)
  loss = criterion(predictions, y_tensor_train)

  loss.backward()

  with torch.no_grad():
    for param in model_v1.parameters():
      if epoch >= 10 :
        print(param)
      param -= lr * param.grad
      param.grad.zero_()

  # Affichage
  if(epoch+1) % 100 ==0 :
    model_v1.eval()

    with torch.no_grad():
      val_pred = model_v1(X_tensor_val)
      val_loss = criterion(val_pred, y_tensor_val)
    print(f"Époque[{epoch+1}/{epochs}] | Erreur train : {loss.item(): .4f} | Erreur val : {val_loss.item(): .4f}")
print("Fin entraînement")

````

# 1.2 MLP optimizer [SGD, SGD + Momentum, Adam] (V2)

````python
for nom_optimizer, set_up in optimizers.items() :#[(SGD,), (SGD+M,), (Adam)]
  model_v2 = copy.deepcopy(model)
  optimizer = set_up(model_v2)

  for epoch in range(epochs):
    model_v2.train()
    predictions = model_v2(X_tensor_train)
    loss = criterion(predictions, y_tensor_train)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    model_v2.eval()
    with torch.no_grad():
        val_pred = model_v2(X_tensor_val)
        val_loss = criterion(val_pred, y_tensor_val)

    # Affichage
    if(epoch+1) % 100 ==0 :
      print(f" Optimizer : {nom_optimizer} | Époque[{epoch+1}/{epochs}] | Erreur train : {loss.item(): .4f} | Erreur val : {val_loss.item(): .4f}")
  print("Fin entraînement")
````

# 1.3.a Configuration 1


````python
model_v3a = nn.Sequential(
    nn.Linear(in_features=61 , out_features=128), # 61 catactéristiques par molécules
    nn.Sigmoid(),

    nn.Linear(in_features=128 , out_features=64),
    nn.Sigmoid(),

    nn.Linear(in_features=64 , out_features=32),
    nn.Sigmoid(),

    nn.Linear(in_features=32, out_features=16),
    nn.Sigmoid(),

    nn.Linear(in_features=16 , out_features=8),
    nn.Sigmoid(),

    nn.Linear(in_features=8, out_features=1),
)


# Initialisation selon Xavier Glorot et Yoshua Bengio
def init_Glorot_Bengio(_model):
  for layer in _model :
    if isinstance(layer, nn.Linear):
      nn.init.xavier_normal(layer.weight)
      nn.init.zeros_(layer.bias)

# Optimiseur
optimizer = optim.Adam(model_v3a.parameters(), lr_opt)
# Application de l'initialisation
init_Glorot_Bengio(model_v3a)

# Boucle d'entraînement et de validation
for epoch in range(epochs):
    model_v3a.train()
    predictions = model_v3a(X_tensor_train)
    loss = criterion(predictions, y_tensor_train)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    model_v3a.eval()
    with torch.no_grad():
        val_pred = model_v3a(X_tensor_val)
        val_loss = criterion(val_pred, y_tensor_val)

    # Affichage
    if(epoch+1) % 100 ==0 :
      print(f" Époque[{epoch+1}/{epochs}] | Erreur train : {loss.item(): .4f} | Erreur val : {val_loss.item(): .4f}")
print("Fin entraînement")
````

````python
model_v3b = nn.Sequential(
    nn.Linear(in_features=61 , out_features=128), # 61 catactéristiques par molécules
    nn.ReLU(),

    nn.Linear(in_features=128 , out_features=64),
    nn.ReLU(),

    nn.Linear(in_features=64 , out_features=32),
    nn.ReLU(),

    nn.Linear(in_features=32, out_features=16),
    nn.ReLU(),

    nn.Linear(in_features=16 , out_features=8),
    nn.ReLU(),

    nn.Linear(in_features=8, out_features=1),
)


# Initialisation selon He
def init_He(_model):
  for layer in _model:
    if isinstance(layer, nn.Linear):
      nn.init.He(layer.weight)
      nn.init.zeros_(layer.bias)

# Optimiseur
optimizer = optim.Adam(model_v3b.parameters(), lr_opt)
# Application de l'initialisation
init_He(model_v3b)

# Boucle d'entraînement et de validation
for epoch in range(epochs):
    model_v3b.train()
    predictions = model_v3b(X_tensor_train)
    loss = criterion(predictions, y_tensor_train)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    model_v3b.eval()
    with torch.no_grad():
        val_pred = model_v3b(X_tensor_val)
        val_loss = criterion(val_pred, y_tensor_val)

    # Affichage
    if(epoch+1) % 100 ==0 :
      print(f" Époque[{epoch+1}/{epochs}] | Erreur train : {loss.item(): .4f} | Erreur val : {val_loss.item(): .4f}")
print("Fin entraînement")
````

### 2.2 Calcul d'attention à la main (dérivation manuscrite)

Considérez le SMILES `CCO` avec $d_{\text{model}} = 4$ et $d_k = 2$. Les matrices de projection sont :

$$W_Q = \begin{pmatrix} 1 & 0 \\ 0 & 1 \\ 1 & 1 \\ 0 & 0 \end{pmatrix}, \quad W_K = \begin{pmatrix} 0 & 1 \\ 1 & 0 \\ 0 & 1 \\ 1 & 0 \end{pmatrix}, \quad W_V = \begin{pmatrix} 1 & 1 \\ 0 & 0 \\ 1 & 0 \\ 0 & 1 \end{pmatrix}$$

Les plongements sont des vecteurs colonnes, et la convention est $q_i = W_Q^\top e_i$, $k_i = W_K^\top e_i$, $v_i = W_V^\top e_i$. De manière équivalente, si l'on empile les plongements comme lignes d'une matrice $E \in \mathbb{R}^{3 \times 4}$, alors $Q = E\,W_Q$, $K = E\,W_K$, $V = E\,W_V$. Les trois plongements sont :

$$e_1 = e_C = \begin{pmatrix} 1 \\ 0 \\ 1 \\ 0 \end{pmatrix}, \quad e_2 = e_C = \begin{pmatrix} 1 \\ 0 \\ 1 \\ 0 \end{pmatrix}, \quad e_3 = e_O = \begin{pmatrix} 0 \\ 1 \\ 0 \\ 1 \end{pmatrix}$$

Calculez à la main : (1) les matrices $Q$, $K$ et $V$, (2) les scores d'attention $QK^\top / \sqrt{d_k}$, (3) les poids après softmax, (4) la sortie finale de la couche d'attention.
